**Step 1 : Data Reading**

```python
import seaborn as sns
df= sns.load_dataset('tips')
df.head()
```

**Step 3 : Data preprocessing (Encoding)**

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['time']=le.fit_transform(df['time'])
```

```python
df.time
```

```python
df.time.unique()
```

**step 4 : Defining independent and dependent variable**

```python
X=df.drop(labels=["time"],axis=1)
y=df.time
```

```python
X.head()
```

```python
y.head()
```

```python
X['day'].value_counts()
```

**step 6 : performing Train Test split**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1)
```

**step 7: Understanding Concept of Piplines**

**Automated Feature Engineering**

```python
df.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
```

```python
catagorical_cols=['sex','smoker','day']
numerical_cols=['total_bill','tip','size']
```

```python
#feature Engineering Automation
#numerical pipline
num_p=Pipeline(
    steps=[
    ('imputer',SimpleImputer(strategy='median')),#missing values
    ('scaler',StandardScaler())#Scaling
])

#catagorical pipeline
cat_p=Pipeline(
    steps=[
    ('imputer',SimpleImputer(strategy='most_frequent')),
    ('onehot',OneHotEncoder())
])
```

```python
#Combining both piplines
preprocessor=ColumnTransformer([
    ('num_pipeline',num_p,numerical_cols),
    ('cat_pipeline',cat_p,catagorical_cols)
])
```

```python
preprocessor
```

```python
preprocessor.fit_transform(X_train)
```

```python
X_train=preprocessor.fit_transform(X_train)
X_test=preprocessor.transform(X_test)
```

**step 8 : Model Traning**

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier()
```

```python
model.fit(X_train,y_train)
```

**step 9 : Prediction**

```python
# To fix this, store predictions alongside the test set targets rather than the original df
test_results = y_test.to_frame(name='actual')
test_results['prediction'] = model.predict(X_test)
test_results.head()
```
