# **Implementation of Support Vector Machine**

SVM works by mapping data to a high-dimensional feature space so that data points can be categorized, even when the data are not otherwise linearly separable (This gets done by kernel function of SVM classifier). A separator between the categories is found, then the data is transformed in such a way that the separator could be drawn as a hyperplane.

The column where datatype is object is non numeric that means we can not apply neumarical operations on these perticular columns

```python
x = df.iloc[:, [2, 3]].values
y = df.iloc[:, 4].values
```

```python
x[:5]
```

**step 4: splitting data into train test split**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=1)
```

**step 5 : Data Scaling**

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
X_train = sc_x.fit_transform(X_train)
X_test = sc_x.transform(X_test)
```

A kernel in a Support Vector Machine (SVM) is a mathematical function that takes low-dimensional input data and transforms it into a higher-dimensional space. Its primary purpose is to help the SVM find a linear decision boundary (a hyperplane) for datasets that are not linearly separable in their original space.

**step 6 : model Training**

```python
from sklearn.svm import SVC
classifier = SVC(kernel='linear', random_state=0)
classifier.fit(X_train, y_train)
```

```python
df['y_pred']= classifier.predict(X_test)
```
