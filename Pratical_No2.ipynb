```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data ={"hours" : [1,2,3,4,5],
       "Marks" : [50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
X= df[["hours"]].values
y=df[["Marks"]].values
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)
```

```python
# Extract slope (coef_) and intercept from the trained model
slope = model.coef_[0]
intercept = model.intercept_

#"f-string" (formatted string literal) in Python
#.2f: This is a format specifier used within the curly braces of an f-string
print(f"slope(m): {slope}")
print(f"Intercept (b): {intercept}")
print(f"Equation of the regression line: y = {slope[0]:.2f}x + {intercept[0]:.2f}")
```

```python
df['Predicted Marks']=model.predict(X)
```

```python
plt.scatter(df['hours'],df['Marks'],color='blue',label='Actual Marks')
```

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label='Regression Line')
plt.xlabel('Hours studied')
plt.ylabel('Marks')
plt.title('Linear Regression:Hours vs Marks')
plt.legend()
plt.show()
```
